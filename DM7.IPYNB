```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\AB\Downloads\archive (20)\heart.csv")

print(df.head())
```

```python
print(df.shape)
```

```python
print(df.isnull().sum())
```

```python
print(df.describe())
```

```python
print("Duplicate Rows:", df.duplicated().sum())

df = df.drop_duplicates()
```

```python
print(df["target"].value_counts())
```

```python
plt.figure(figsize=(6,4))
sns.histplot(df["age"], bins=20)
plt.title("Age Distribution")
plt.show()
```

```python
print("Average Blood Pressure:", df["trestbps"].mean())
```

```python
plt.figure(figsize=(5,4))
sns.countplot(x="sex", data=df)
plt.title("Gender Distribution")
plt.show()
```

```python
plt.figure(figsize=(5,4))
sns.countplot(x="target", data=df)
plt.title("Heart Disease Distribution")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.histplot(df["chol"], bins=20)
plt.title("Cholesterol Distribution")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.histplot(df["trestbps"], bins=20)
plt.title("Resting Blood Pressure")
plt.show()
```

```python
pd.crosstab(df["sex"], df["target"])
```

```python
sns.countplot(x="sex", hue="target", data=df)
plt.title("Heart Disease by Gender")
plt.show()
```

```python
print("\n----- Medical Insights -----")

print("Total Patients:", len(df))

print("Patients with Heart Disease:",
      (df["target"] == 1).sum())

print("Patients without Heart Disease:",
      (df["target"] == 0).sum())

print("Average Age:",
      round(df["age"].mean(),2))

print("Average Cholesterol:",
      round(df["chol"].mean(),2))

print("Average Blood Pressure:",
      round(df["trestbps"].mean(),2))
```
